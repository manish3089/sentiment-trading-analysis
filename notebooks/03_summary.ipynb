

## Methodology

1. **Data Preparation**  
   - Collected daily trader metrics (PnL, win rate, trade counts, trade sizes).  
   - Aggregated at the **account–day level** to normalize outcomes across traders.  
   - Cleaned and structured datasets with explicit naming conventions for reviewer clarity.

2. **Trader Classification**  
   - **Frequency**: Classified traders as *Frequent* or *Infrequent* based on daily trade counts (above/below median).  
   - **Consistency**: Defined categories (*Consistent, Inconsistent, Mixed*) using thresholds:  
     - Win rate >= median and PnL volatility <= median = Consistent  
     - Win rate < median and PnL volatility > median = Inconsistent  
     - Otherwise = Mixed

3. **Aggregation & Analysis**  
   - Computed account‑level averages: daily PnL, win rate, PnL standard deviation, and average trade size.  
   - Compared groups using bar plots with error bars to visualize mean differences and variability.  
   - Highlighted behavioral contrasts:  
     - Frequent vs Infrequent : activity vs stability trade‑off.  
     - Consistent vs Inconsistent : risk‑adjusted discipline vs volatility‑driven outcomes.

4. **Insights Extraction**  
   - Found that **Inconsistent traders** show higher average PnL due to larger trade sizes and higher variance, but at the cost of stability.  
   - **Consistent traders** deliver lower average PnL but steadier outcomes, reflecting disciplined risk management.  
   - **Frequent traders** outperform in average daily PnL but with greater exposure to noise, while **Infrequent traders** trade selectively with steadier win rates.

# Insights
- Reversal bias is costly in fear phases: Traders increase long positions during Fear and Extreme Fear, but this coincides with the highest drawdowns, indicating that buying dips in volatile markets leads to poorer outcomes.
- Higher trade frequency does not improve accuracy: Despite a significant increase in the number of trades during Extreme Fear, win rate remains unchanged, suggesting that additional trades add noise rather than value.
- Larger position sizes drive both gains and losses: Inconsistent traders achieve higher average PnL primarily due to larger trade sizes, but this also results in greater volatility and risk exposure.
- Trading activity increases risk without improving returns: Frequent traders generate more trades and wider PnL distributions, yet do not outperform, indicating diminishing returns from increased activity.

# Part C — “Actionable output”

# Strategy 1: Focus on Trend-Following in Fear Phases

During Fear and Extreme Fear, traders show a stronger long bias, indicating a tendency to trade reversals (buying dips) rather than following the prevailing downward trend. However, these periods also exhibit the highest drawdowns, suggesting that reversal trading in volatile markets is risky.

Traders should instead focus on aligning with the trend rather than attempting to catch reversals.

# Strategy 2: Reduce Trade Frequency in Extreme Fear

During Extreme Fear, traders should reduce the number of trades and be more selective. The analysis shows that trade frequency is highest in these periods, which contributes to greater volatility in PnL without improving win rates.

Limiting the number of trades and focusing only on high-quality setups can help reduce unnecessary exposure and stabilize performance.
